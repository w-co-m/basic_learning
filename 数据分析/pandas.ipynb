###一.pandas库

```python
import pandas as pd
#series 的创建
s = pd.Series([1, 3, 5, 7, 9])
print(s)
#自定义索引
s = pd.Series([1, 3, 5, 7, 9], index=['a', 'b', 'c', 'd', 'e'])
print(s)
#自定义name
s = pd.Series([1, 3, 5, 7, 9], index=['a', 'b', 'c', 'd', 'e'], name='my_series')
print(s)
```

```python
#通过字典创建Series
data = {'a': 1, 'b': 3, 'c': 5, 'd': 7, 'e': 9}
s = pd.Series(data, name='my_series')
print(s)
print(s.values)
print(s.index)
print(s.name, s.dtype, s.shape, s.ndim)
print(s.loc['c'])  #通过标签访问
print(s.iloc[2:4])   #通过位置访问
```

```python
#访问数据
print(s['c'])  #通过标签访问
print(s[s<5])  #通过条件访问
print(s[1:4])  #通过位置访问
print(s.iloc[1:4])  #通过位置访问
print(s.loc['b':'d'])  #通过标签访问    
print(s.loc['b':'d':2])  #通过标签访问，步长为2
print(s.head(3))  #访问前3个元素
print(s.tail(2))  #访问后2个元素
```

```python
#常用方法
print(s.sum())  #求和
print(s.mean())  #求平均值
print(s.std())  #求标准差
print(s.var())  #求方差
print(s.min())  #求最小值
print(s.max())  #求最大值
print(s.median())  #求中位数
print(s.mode())  #求众数
print(s.describe())  #描述统计
print(s.value_counts())  #统计每个值出现的次数
print(s.sort_values())  #按值排序
print(s.sort_index())  #按索引排序
print(s.unique())  #返回唯一值
print(s.isnull())  #判断是否为缺失值
print(s.notnull())  #判断是否为非缺失值
print(s.dropna())  #删除缺失值
print(s.fillna(0))  #填充缺失值为0
print(s.astype(float))  #转换数据类型为float
print(s.apply(lambda x: x**2))  #对每个元素应用函数，这里是求平方  
```

```python
'''
每小时销售额数据分析
商店每小时销售额为Series
按天重采样计算每日总销售额
计算每天营业时间(8:00-22:00)和非营业时间的销售额比例
找出销售额最高的三个小时
'''
#创建每小时销售额数据
import numpy as np
data=pd.Series(np.random.randint(0,100,size=24),
               index=pd.date_range('2024-01-01 00:00',periods=24,freq='h'),name='sales')
print(data)
print("\n每日总销售额:")
#按天重采样计算每日总销售额
daily_sales=data.resample('D').sum()
print(daily_sales)
print("\n每天营业时间和非营业时间的销售额比例:")
#计算每天营业时间(8:00-22:00)和非营业时间的销售额比例
business_hours_sales=data.between_time('8:00','22:00').sum()
non_business_hours_sales=data.between_time('22:00','8:00').sum()
sales_ratio=business_hours_sales/(business_hours_sales+non_business_hours_sales)
print(sales_ratio)
print("\n销售额最高的三个小时:")
#找出销售额最高的三个小时
top_hours=data.nlargest(3)
print(top_hours)
```
